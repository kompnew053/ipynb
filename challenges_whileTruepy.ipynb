```python
print('===Apakah Kita Sudah Sampai?==')
print('\nJika sudah sampai, masukkan "sudah sampai"')
while True:
  i = input('\nApakah sudah sampai? ')
  if i == 'sudah sampai':
    break
```

```python
print('===Are we there yet??==')
print('\nJika sudah sampai, masukkan "Yes"\n')

while True:
  answer = input('Are we there yet? ')
  if answer.lower() != 'yes':
    answer = input('Are we there yet? ')
  elif answer.lower() == 'yes':
    break
```

```python
import random

while True:
  dadu1 = random.randint(1,6)
  dadu2 = random.randint(1,6)
  total = dadu1 + dadu2
  if total != 2:
    print('Nope')
    print('rerolling')
  elif total == 2:
    print("Snake eyes!")
```

```python
print('               ===Jumlah Kuadrat=== ')
print('Program ini akan menampilkan hasil jumlah kuadrat bilangan \ndari bilangan 1 hingga bilangan yang ditentukan')

number = int(input('\nMasukkan bilangan: '))

total=0
for i in range(1, number+1):
  square = i**2
  total = total+square
print(f'Hasil: {total}')
```
