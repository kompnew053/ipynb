```python
import sympy as sp

x = sp.Symbol('x', real=True)
expr1 = (2*x + 1)/(x - 2) - (x - 3)/(x + 1)
expr2 = (2*x + 1)/(x - 2) + (x - 3)/(x + 1)

expr1_simplified = sp.simplify(expr1)
expr2_simplified = sp.simplify(expr2)

print("Part 1:", expr1_simplified)
print("Part 2:", expr2_simplified)

full_expr = expr1_simplified * expr2_simplified
full_expr_simplified = sp.simplify(full_expr)
print("Full:", full_expr_simplified)

sol = sp.solve_univariate_inequality(sp.Abs((2*x+1)/(x-2)) > sp.Abs((x-3)/(x+1)), x, relational=False)
print("Solution:", sol)
```
