```python
# @title
import time

bold = "\033[1m"       # Tebal
italic = "\033[3m"     # Miring
reset = "\033[0m"

print(bold+'                       *~~Halo!~~*'+reset)
print(bold+'             SELAMAT DATANG DI KUIS ENZIM'+reset)
print(italic+'by: Gina Fitriani'+reset)
print(' ')
print('🌟Di sini akan ada 10 pertanyaan.')
print('🌟Setiap 1 soal yang benar akan mendapatkan 10 poin!')
print('🌟Jawab setiap pertanyaan dengan benar yaa~!')
print(' ')

high_score = 0
shortest_time = float('inf')

while True:
    print(bold+"\n=== Menu Kuis ==="+reset)
    print("1. Mulai Kuis")
    print("2. Lihat High Score")
    print("3. Keluar")

    pilihan = input("→ Pilih menu (1-3): ")

    while pilihan not in ["1", "2", "3"]:
        print(' ')
        print(bold+"Pilihan tidak valid. Silakan pilih menu yang tersedia."+reset)
        pilihan = input("→ Pilih menu (1-3): ")

    print(' ')
    print(bold+"="*60+reset)
    print(' ')

    if pilihan == "1":
        score = 0
        start_time = time.time()   # mulai hitung waktu

        print('1️⃣ Mempercepat jalannya reaksi kimia.')
        print('    a. Metabolisme')
        print('    b. Biokatalisator')
        print('    c. Enzim')
        jawaban = input('→ jawaban: ')
        if jawaban.lower() == "b":
            print("✅Yeayy !Benar!!")
            score += 10
        else:
            print("❌Yah, jawabannya: b. Biokatalisaror!")

        print(' ')
        print('')
        print('2️⃣ Bagian enzim yang berfungsi sebagai tempat menempelnya substrat?')
        print('    a) Sisi aktif')
        print('    b) alosterik')
        print('    c) enzim')
        jawaban = input('→ jawaban: ')
        if jawaban.lower() == "a":
            print("✅Yeayy! Benar!")
            score += 10
        else:
            print("❌Yah, jawabannya a. Sisi aktif!")

        print(' ')
        print(' ')
        print('3️⃣ Tokoh teori lock and key.')
        print('    a) Daniel Koshland')
        print('    b) Koshland Emil')
        print('    c) Emil Fischer')
        jawaban = input('→ jawaban: ')
        if jawaban.lower() == "c":
            print("✅Yeayy! Benar!!")
            score += 10
        else:
            print("❌Yah, jawabannya c. Emil Fischer.")

        print(' ')
        print('')
        print('4️⃣ Senyawa protein yang berfungsi sebagai biokatalisator, dengan cara menurunkan energi aktivasi.')
        print("    a) Biokatalisator")
        print("    b) Apoenzim")
        print("    c) Enzim")
        jawaban = input('→ jawaban: ')
        if jawaban.lower() == "c":
            print("✅Yeayy! Benar!!")
            score += 10
        else:
            print("❌Yah, jawabannya c. Enzim!")

        print(' ')
        print(' ')
        print('5️⃣ Molekul yang diikat oleh enzim disebut?')
        print('    a) Substrat')
        print('    b) Koenzim')
        print('    c) Aktivator')
        jawaban = input('→ jawaban: ')
        if jawaban.lower() == "a":
            print('✅Yeayy! Benar!!')
            score += 10
        else:
            print("❌Yah, jawabannya a. Substrat!")

        print(' ')
        print(' ')
        print('6️⃣ Enzim termasuk ke dalam golongan?')
        print('    a) Karbohidrat')
        print('    b) Lemak')
        print('    c) Protein')
        jawaban = input('→ jawaban: ')
        if jawaban.lower() == "c":
            print('✅Yeayy! Benar!!')
            score += 10
        else:
            print("❌Yah, jawabannya c. Protein!")

        print(' ')
        print(' ')
        print('7️⃣ Enzim amilase berfungsi mengubah?')
        print('    a) Protein menjadi asam amino')
        print('    b) Amilum menjadi maltosa')
        print('    c) Sukrosa menjadi glukosa')
        jawaban = input('→ jawaban: ')
        if jawaban.lower() == "b":
            print('✅Yeayy! Benar!!')
            score += 10
        else:
            print("❌Yah, jawabannya b. Amilum menjadi maltosa!")


        print(' ')
        print(' ')
        print('8️⃣ Enzim bekerja secara spesifik, artinya?')
        print('    a) Hanya bekerja pada pH tertentu')
        print('    b) Hanya bekerja pada suhu tertentu')
        print('    c) Hanya dapat mengikat substrat tertentu')
        jawaban = input('→ jawaban: ')
        if jawaban.lower() == "c":
            print('✅Yeayy! Benar!!')
            score += 10
        else:
            print("❌Yah, jawabannya c. Hanya dapat mengikat substrat tertentu!")


        print(' ')
        print(' ')
        print('9️⃣ Faktor yang dapat merusak struktur enzim sehingga enzim tidak aktif lagi adalah?')
        print('    a) Suhu terlalu tinggi')
        print('    b) Jumlah substrat banyak')
        print('    c) Jumlah produk sedikit')
        jawaban = input('→ jawaban: ')
        if jawaban.lower() == "a":
            print('✅Yeayy! Benar!!')
            score += 10
        else:
            print("❌Yah, jawabannya a. Suhu terlalu tinggi!")


        print(' ')
        print(' ')
        print('🔟 Enzim katalase banyak terdapat pada organ?')
        print('    a) Ginjal')
        print('    b) Hati')
        print('    c) Lambung')
        jawaban = input('→ jawaban: ')
        if jawaban.lower() == "b":
            print('✅Yeayy! Benar!!')
            score += 10
        else:
            print("❌Yah, jawabannya b. Hati!")


        # Hitung lama waktu
        end_time = time.time()
        total_time = round(end_time - start_time, 2)

        print(' ')
        print(bold+"="*60+reset)
        print(' ')
        #Skor akhir
        print(f"🌟Skor akhir kamu: {score}/100")
        print(f"⏱️ Waktu yang kamu habiskan: {total_time} detik")


        if score > high_score:
            high_score = score
            shortest_time = total_time
            print(bold+"🎉 Selamat! Anda mendapatkan High Score baru! 🎉"+reset)

        print(' ')
        print(bold+"="*60+reset)


    elif pilihan == "2":
        if high_score > 0:
            print(bold+f"High Score: {high_score}/100"+reset)
            print(bold+f"Waktu untuk High Score ini: {shortest_time} detik"+reset)
        else:
            print(bold+"Belum ada High Score yang tercatat."+reset)

        print(" ")
        print(bold+"="*60+reset)
    elif pilihan == "3":
        score = 0
        print(bold+"           ---Terima kasih telah bermain!---"+reset)
        break
```
